# [exploring]

```python
# -select only the important columns

df = song[['Rank', 'Artist', 'Track Name', 'Artist_popularity', 'Streams', 'tempo', 'danceability', 'energy']]
df
```

```python
# -get the Artist_popularity which are greater that 90
pop_song = df[df['Artist_popularity'] > 90]
pop_song
```

```python
# -artist with most songs in the list, by count how frequent of the artist name appeared in the data frame
rank = pop_song["Artist"].value_counts()
rank
```

A function that calculates the correlation between pairs of columns, excluding NA/null values. It has the following three methods for the 'method' parameter:
- pearson: standard correlation coefficient
- kendall: Kendall Tau correlation coefficient
- spearman: Spearman rank correlation

```python
cor = pop_song.corr(method='pearson', numeric_only=True)
cor
```

```python
plt.matshow(cor)
plt.colorbar()
```
